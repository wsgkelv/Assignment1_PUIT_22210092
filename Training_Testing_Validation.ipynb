```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Dummy dataset
data = pd.DataFrame({
    "age": [12, 15, 18, 21, 25, 30, 34, 40, 50, 60],
    "height": [140, 150, 160, 170, 175, 178, 180, 182, 170, 165]
})

# Features and label
X = data[["age"]]         # Independent variable (feature)
Y = data["height"]        # Dependent variable (label/target)

# 🔹 FIRST SPLIT → Train (70%) + Temp (30%)
X_train, X_temp, Y_train, Y_temp = train_test_split(
    X, Y,
    test_size=0.3,        # 30% of data is set aside for temp
    random_state=42
)

# 🔹 SECOND SPLIT → Temp divided into Validation (15%) + Test (15%)
X_val, X_test, Y_val, Y_test = train_test_split(
    X_temp, Y_temp,
    test_size=0.5,        # Half of temp → 15% of total becomes test
    random_state=42
)

# Showing the size of each split:
print("Training Size:", len(X_train))
print("Validation Size:", len(X_val))
print("Testing Size:", len(X_test))

# Optional Display:
print("\nTraining Set:")
display(pd.concat([X_train, Y_train], axis=1))

print("\nValidation Set:")
display(pd.concat([X_val, Y_val], axis=1))

print("\nTesting Set:")
display(pd.concat([X_test, Y_test], axis=1))
```
